Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng

W6A5 
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 
'zeros': ...} (đếm số >0, <0, =0) 

```python
num=list(map(int,input().split()))
countpos=0
countneg=0
countzero=0
res=[]
for i in range(len(num)):
    if num[i]>0:
        countpos += 1
        res.append("positive")
    elif num[i]==0:
        countzero += 1
        res.append("zero")
    else:
        countneg += 1
        res.append("negative")
```

**W6A9 
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một 
dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ 
được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới 
theo thứ tự từ điển tăng dần của key.**

```python
#example input: {"a":1, b":2}}  etc n remember the {}
dict1=eval(input())
dict2=eval(input())
res={}

res=dict1.copy()
for key in dict2:
    if key in res:
        res[key] = res[key] + dict2[key]
    else:
        res[key] = dict2[key]

sorted_keys = sorted(res.keys())

fin={}
for key in sorted_keys:
    fin[key] = res[key]
print(fin)
```

W6A12 
[Tìm phần tử xuất hiện nhiều nhất] 
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong 
list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ 
nhất trong số đó. 

```python
lit=list(map(int,input().split()))
check=[]
count={}
max=0
mostapp= None
for i in lit:
    if i not in check:
        check.append(i)
        count[i] = 1
    else:
        count[i] += 1
    
    current=count[i]
    if current>max:
        max=current
        mostapp=i
    elif current==max:
        if mostapp is None or current<mostapp:
            mostapp=i
print(mostapp)
```

W6A15 [Lọc dictionary theo giá trị] 
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số 
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp 
key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary 
mới.


```python
#note:  key is string and value is int , remember the {}
ogdict= eval(input())
k=int(input())
res={}
for i in ogdict:
    if ogdict[i]>k:
        res[i]=ogdict[i]
print(res)
```
